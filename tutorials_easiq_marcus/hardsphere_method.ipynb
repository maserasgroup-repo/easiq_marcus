# 1 -  Estimate the energy barrier of SET/EnT process using Hard-sphere and simplified Hard-sphere methods

### In this tutorial, we demonstrated how to estimate the energy barrier of SET/EnT process by applying the Hard-sphere and simplified Hard-Sphere methods.  
These methods provide a quick and empirical way to estimate energy barriers without extensive electronic structure calculations. For more advanced calculations, refer to the package documentation and explore other methods available.

In this example, the reaction corresponds to the following SET process: \
Ni(II) + *[Ir(III)] → [Ni (III)]$^+$  + [Ir(II)] \
where the files are: \
ni_2.out ir_3_exc.out ni_3.out ir_2.out

```python
!python -m eqm.get_results_feb25 --help
```

### As detailed in the --help message, there are multiples parameters to specify. Let's break them down...
### To begin with the simplified Hard-Sphere model, we need to: (i) select the A constant, and (ii) provide either the volume of the donor and acceptor reactants, or the radius of the donor and acceptor reactants.
NOTE: this information can be obtained in the Gaussian output file by including the keyword "volume" in the input line.

```python
radius_ni = 8.3  # radius of the acceptor
radius_ir = 7.1  # radius of the donor 
```

```python
!python -m eqm.get_results_feb25 -r ni_2.out ir_3_exc.out -p ni_3.out ir_2.out -shs 95 -r_d 7.1 -r_a 8.3
```

### In the Hard-sphere model is preferred, it is also necessary to specify the static and optical dielectric constants of the solvent. 
NOTE: the optical dielectric constant can be calculated by squaring the refractive index of the solvent, i.e. $$
  \varepsilon(\infty) = \varepsilon(\omega = \infty) \approx n^2
$$

```python
d_static = 37.78  # static dielectric constant of the DMA
d_optical = 2.07  # optical dielectric constant of the DMA
```

```python
!python -m eqm.get_results_feb25 -r ni_2.out ir_3_exc.out -p ni_3.out ir_2.out -hs -r_d 7.1 -r_a 8.3 -d 37.78 -d_opt 2.07
```

### If the Saveànt's model is required due to the bond cleavage during the SET/EnT process, the energy barrier can also be estimated by specifying the bond dissociation energy (BDE) of the cleaved bond. ###

In this case, the reaction corresponds to this OSET process: \
Ni(0) + PhBrCF$_3$ → [Ni(I)]$^+$  + PhCF$_3$ + Br$^.$ \
BDE(PhCF$_3$-Br) = G(PhBrCF$_3$) - G(PhCF$_3$) - G(Br$^.$)  \
the files are: \
M0-c.out phbrcf3.out M0-c_I.out phcf3.out br.out

```python
!python -m eqm.get_results_feb25 -r M0-c.out phbrcf3.out -p M0-c_I.out phcf3.out br.out -hs -r_d 8.3 -r_a 6.0 -d 37.78 -d_opt 2.07 -bde 69.2
```
