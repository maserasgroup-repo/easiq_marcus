# 2 -  Estimate the energy barrier of SET/EnT process using the 4-points DFT approach of Marcus theory

### In this tutorial, we will demonstrate how to correctly structure the command line to generate the necessary files for estimating the energy barrier of SET/EnT process using 4-points DFT approach of Marcus theory. We will explore two different formats for representing the energy profiles of the reactants and products: symmetric and asymetric parabolas.
###  There are several steps to follow. 

For this tutorial, we consider the following SET process: \
Ni(II) + *[Ir(III)] → [Ni (III)]+  + [Ir(II)] \
the files are: \
ni_2.out ir_3_exc.out ni_3.out ir_2.out

## 2.1. Run the `get_info` script

### The first step is to create a dictionary.

```python
!python -m eqm.get_info --help
```

### As detailed in the `--help` command, reactants, products, solvent and solvent model must be specified. In this case, the solvent is `n,n-DiMethylAcetamide` and the solvent model is `pcm`.

### If `--chk` is not defined, the script will generate four input files `_solvent.in` to run in order to obtain four `_solvent.chk` files.

NOTE: if the input files contain the basis set (and pseudopotentials) specified in the last line, you will need to introduce Tail argument (or Tail1 and/or Tail2). Otherwise, the script will stop and request the required tail file, as shown in the following example.

```python
!python -m eqm.get_info ni_2.out ir_3_exc.out ni_3.out ir_2.out n,n-DiMethylAcetamide pcm 
```

```python
!python -m eqm.get_info ni_2.out ir_3_exc.out ni_3.out ir_2.out n,n-DiMethylAcetamide pcm --tail1 tail_all
```

### Four new input files are generating with the suffix `_solvent.in`, and the text file `InputstoSumbit.txt` is created. This text file contains the path to the files to run in order to obtain the corresponding `.chk` files for the continue with the 4-points method. ### 

## 2.1. Run the `4points` script to generate the output files required

### Once the four `.chk`files are with the suffix `_solvent.chk`, the script 4points.py has to be executed.

```python
!python -m eqm.4points --help
```

### As detailed in the --help command, reactants, products, solvent and solvent model must be specified. As before, the solvent is `n,n-DiMethylAcetamide` and the solvent model is `pcm`.¶

### If `--usebasisset` is not specified (or set to `--usebasisset no`), the script will automatically retrieve the basis set from the `_solvent.chk` files. Conversely, if  `--usebasisset yes`, the basis set will be retrieved from the dictionary named `input_data.pkl`.

NOTE: if the input files contain the basis set (and pseudopotentials) specified manually in the bottom of the input and `--usebasisset yes` is set. the script will automatically set  `--usebasisset no`. 

```python
!python -m eqm.4points ni_2.out ir_3_exc.out ni_3.out ir_2.out n,n-DiMethylAcetamide pcm --usebasisset yes
```

```python
!python -m eqm.4points ni_2.out ir_3_exc.out ni_3.out ir_2.out n,n-DiMethylAcetamide pcm --usebasisset no
```

## 0.3. Run the `get_results` script to estimate the energy barrier using the 4-points method of Marcus theory

```python
!python -m eqm.get_results_feb25 --help
```

### After performing all these calculations, you will have the following files `ir_2_noneq.out, ir_3_exc_noneq.out, ni_2_noneq.out, ni_3_noneq.out` along with the corresponding eight `.chk` files, two per compound.

### Use the `-fp` flag to select the 4-points method considering symmetric parabolas.

```python
!python -m eqm.get_results_feb25 -r ni_2.out ir_3_exc.out -p ni_3.out ir_2.out -fp
```

### If the reorganization terms are significantly different, use the `-fpa` flag to apply the 4-points method considering asymmetric parabolas.

```python
!python -m eqm.get_results_feb25 -r ni_2.out ir_3_exc.out -p ni_3.out ir_2.out -fpa
```
